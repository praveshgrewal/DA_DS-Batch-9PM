```python
import timeit

list_time = timeit.timeit("[1,2,3,4,5]", number=1000000)
tuple_time = timeit.timeit("(1,2,3,4,5)", number=1000000)

print("List:", list_time)
print("Tuple:", tuple_time)
```

```python
import sys

lst = [1,2,3,4,5]
tup = (1,2,3,4,5)

print(sys.getsizeof(lst))
print(sys.getsizeof(tup))
```
