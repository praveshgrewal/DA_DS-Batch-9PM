### string ---> text

```python
# string ---> "", ''.""" """, ''' '''
name = "suhani"
print(name)
print(type(name)) # to find the type of variable (data type)
```

```python
surname = 'rohilla'
print(surname)
print(type(surname))
```

```python
age = """17"""
print(age)
print(type(age))
```

### int ---> number

```python
age = 17 # whole number
print(age)
print(type(age))
```

### float ---> decimal values

```python
score = 70.6
print(score)
print(type(score))
```

### boolean(bool) ---> True/False

```python
f1 = True # yes,1
print(f1)
print(type(f1))
```

```python
f2 = False # no,0
print(f2)
print(type(f2))
```

### None --- >nothing
* as a placeholder

```python
no = None
print(no)
print(type(no))
```

### complex number

```python
# real + imaginary number
c1 = 2 + 3j
print(c1)
print(type(c1))
```

### list ---> [ ]

```python
my_list = ["suhani",17,70.6,True]
print(my_list)
print(type(my_list))
```

### tuple ---> ( )

```python
my_tuple = ("suhani",17,70.6,True)
print(my_tuple)
print(type(my_tuple))
```

### set --- { }

```python
my_set = {"suhani",17,70.6,True}
print(my_set)
print(type(my_set)) 
```

### dictionary ---> {key:value}

```python
my_dict = {"name":"suhani","age":17,"score":70.6,"is_student":True}
print(my_dict)
print(type(my_dict))
```

### type conversion

```python
num = 10 # int
print(num)
print(type(num))
num2 = 10.5 # float
print(num2)
print(type(num))

result  = num + num2
print(result)
print(type(result))
```

```python
num5 = "100"
print(num5)
print(type(num5))

# int 
num6 = int(num5)
print(num6)
print(type(num6))

# float
num7 = float(num5)
print(num7)
print(type(num7))

# list
my_list2 = list(num5)
print(my_list2)
print(type(my_list2))
```
