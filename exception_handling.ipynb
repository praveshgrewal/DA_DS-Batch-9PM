```python
try:
    10/"suhani"
except ZeroDivisionError:
    print("You can't divide by zero!")
except TypeError:
    print("You can't divide by string!")
# finally:
#     print("This will always execute")


def fact(num):
    if num ==0 or num ==1:
        return 1
    else:
        return num * fact(num-1)

print(fact(5))
```
