### jumping statments
* break --> will break your loop
* continue ---> will skip current iteration
* pass ---> used as a placeholder

````python
nums = int(input("Enter a number: "))
for i in range(1,11):
    num2 = nums * i
    if num2 == 18:
        break
    print(num2)
    
````

````python
nums = int(input("Enter a number: "))
for i in range(1,11):
    if i == 6:
        break
    print(i*nums)
````

````python
nums = int(input("Enter a number: "))
for i in range(1,11):
    if i == 6:
        continue # skip the iteration
    print(i*nums)
````

````python
# we have to skip even values
nums = [1,7,9,23,45,67,22,44,33,66]
for n in nums:
    if n%2 != 0:
        continue
    print(n)
    
````

## while loop
* true ---> execution
* false ---> stop

````python
# 1, 10
i = 1 # starting point
while i <= 10: # condition 
    print(i) # print(i)
    i += 1 # increment
````

````python
table = int(input("Enter a number: ")) # 2
n = 1
while n <= 10:
    if n == 6:
        break
    print(f"{n} x {table} = {n*table}")
    n += 1
````

````python
# 1 --20
i = 2
while i <= 20:
    print(i)
    i += 2
````
