```python
import numpy as np
import time
import sys

N = 1_000_000

# -------------------------
# Create data
# -------------------------
py_list = list(range(N))
np_array = np.arange(N)

# -------------------------
# SPACE
# -------------------------
print("Space:")
print("Python List :", sys.getsizeof(py_list), "bytes")
print("NumPy Array :", np_array.nbytes, "bytes")


# -------------------------
# TIME: Addition
# -------------------------

# Python List
start = time.perf_counter()

list_result = [x + 100 for x in py_list]

list_time = time.perf_counter() - start


# NumPy Array
start = time.perf_counter()

numpy_result = np_array + 100

numpy_time = time.perf_counter() - start


print("\nTime for adding 10 to 1 million elements:")
print("Python List :", list_time)
print("NumPy Array :", numpy_time)

print("\nNumPy is approximately",
      round(list_time / numpy_time, 2),
      "times faster")
```
